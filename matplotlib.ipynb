```python
# matplotlib basics
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from random import sample
from random import sample
```

```python
x = np.linspace(0,5,11)
print(x)
y = x ** 3
print(y)
```

```python
plt.plot(x,y)
plt.title("title")
plt.xlabel("X axis")
plt.ylabel(" Y axis")
```

```python
# subplot basics
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(y,x)
```

```python
# object oriented way
```

```python
fig = plt.figure(figsize=(20,8),dpi = 500)
axis1 = fig.add_axes([0.1,0.1,.4,.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.2,0.3,0.4])
axis2.plot(x,y)
```

```python
#Types of plots
```

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
data = sample(range(1,1000),100)
plt.hist(data)
```

```python
plt.boxplot(x)
```

```python
# saving plots
```

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```

```python
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)


# Create the plot
plt.figure(figsize=(8, 5))  # Optional: set figure size

# Plot with custom design options
plt.plot(
    x, y1,
    label='Sine Wave',
    color='royalblue',
    linestyle='--',
    linewidth=2,
    marker='o',
    markersize=6,
    markerfacecolor='white',
    markeredgewidth=1.5
)



# Optional design elements
plt.title("Sine vs Cosine Waves", fontsize=16, fontweight='bold', color='navy')
plt.xlabel("X Axis (radians)", fontsize=12)
plt.ylabel("Y Axis (value)", fontsize=12)

# Show the plot
plt.show()
```
