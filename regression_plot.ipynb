```python
# regression plot
# used in machine learning - lmplot()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
sns.lmplot(x = "total_bill",y ="tip",data = tips )
# linear regression algorithm
```

```python
sns.lmplot(x = "total_bill",y = "tip",data = tips,hue = "sex")
```

```python
sns.lmplot(x = "total_bill",y = "tip",data = tips,hue = "sex",palette = "rainbow")
```

```python
sns.lmplot(x = "total_bill",y = "tip",data = tips,hue = "sex",palette = "rainbow",markers = "v")
```

```python
sns.lmplot(x = "total_bill",y = "tip",data = tips,hue = "sex",palette = "rainbow",markers =["v","o"])
```

```python
sns.lmplot(x = "total_bill",y = "tip",data = tips,hue = "sex",palette = "rainbow",markers =["v","o"],scatter_kws = {"s":100})
```
